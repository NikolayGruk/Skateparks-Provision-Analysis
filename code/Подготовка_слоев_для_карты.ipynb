```python
# Импортируем необходимые библиотеки

import pandas as pd
import numpy as np
import geopandas as gpd
import os

# Определяем путь к файлами

data_path = '/content/data'
```

```python
# Загружаем базовые слои

skateparks = gpd.read_file(os.path.join(data_path, 'skateparks.geojson')) # слой со скейт-парками
buildings = gpd.read_file(os.path.join(data_path, 'buildings.geojson')) # слой с жилыми домами
bounds = gpd.read_file(os.path.join(data_path, 'bounds.geojson')) # слой с границами муниципальных образований
```

```python
# Работа со слоем скейт-парков

skateparks = skateparks.rename(columns={
    'TableID': 'Номер в таблице',
    'Name': 'Название',
    'Rating': 'Рейтинг',
    'capacity': 'Вместимость',
    'Administrative': 'Административный район',
    'Munitipal': 'Муниципальное образование',
    'Address': 'Адрес',
    'Operator': 'Обслуживающая организация',
    'Year': 'Год постройки',
    'Area': 'Площадь постройки',
    'Dimensional_type': 'Размерный тип',
    'Construction_type': 'Тип конструкции',
    'Base_material': 'Материал основания',
    'Terms_of_use': 'Правила пользования',
    'Dimensional_solution': 'Объемно-пространственное решение',
    'Disciplines': 'Подходящие дисциплины',
    'Riding_Styles': 'Подходящие стили катания',
    'Styles_count': 'Количество подходящих стилей',
    'Required_riding_level': 'Требуемый уровень катания'
    }, inplace=False)

skateparks.to_file(os.path.join(data_path,"skateparks.geojson"))
skateparks.to_file(os.path.join(data_path,"скейт-парки.geojson"))
```

```python
# Работа со слоем жилых зданий

buildings = buildings.rename(columns={
    'building': 'Тип здания',
    'Administrative': 'Административный район',
    'Munitipal': 'Муниципальное образование',
    'addr:street': 'Улица',
    'addr:housenumber': 'Номер дома',
    'area': 'Площадь постройки',
    'storeys_count': 'Количество этажей',
    'living_area': 'Жилая площадь',
    'population': 'Количество жителей',
    'demand_30': 'Спрос в границах временной доступности 30 минут',
    'demand_60': 'Спрос в границах временной доступности 60 минут',
    'demand_90': 'Спрос в границах временной доступности 90 минут',
    'demand': 'Суммарный спрос'
    }, inplace=False)

buildings.to_file(os.path.join(data_path,"buildings.geojson"))
buildings.to_file(os.path.join(data_path,"жилые здания.geojson"))
```

```python
# Работа со слоем границ муниципальных образований

bounds = bounds[['Administrative', 'Munitipal', 'Population',
       'demand_30', 'demand_60', 'demand_90', 'total_demand', 'skateparks_count', 'geometry']]

bounds = bounds.rename(columns={
    'Administrative': 'Административный район',
    'Munitipal': 'Муниципальное образование',
    'Population': 'Население',
    'demand_30': 'Спрос в границах временной доступности 30 минут',
    'demand_60': 'Спрос в границах временной доступности 60 минут',
    'demand_90': 'Спрос в границах временной доступности 90 минут',
    'total_demand': 'Суммарный спрос',
    'skateparks_count': 'Количество скейт-парков'
    }, inplace=False)

bounds.to_file(os.path.join(data_path,"bounds.geojson"))
bounds.to_file(os.path.join(data_path,"муниципальные образования.geojson"))
```

```python
# Загружаем слои с расчитанными показателями в границах 30 минут

build_provision_30 = gpd.read_file(os.path.join(data_path, 'build_prov_exp_30.geojson')) # показатель обеспеченности жилого дома скейт-парками
distance_between_30 = gpd.read_file(os.path.join(data_path, 'links_prov_exp_30.geojson')) # перемещения между жилыми домами и скейт-парками
skateparks_load_30 = gpd.read_file(os.path.join(data_path, 'services_prov_exp_30.geojson')) # показатели транспортной доступности и загруженности скейт-парков
final_assessment_30 = gpd.read_file(os.path.join(data_path, 'exp1_30_mun (на русском).geojson')) # аггрегация результатов по муниципальным округам
```

```python
# Показатели обеспеченности жилого дома скейт-парками

build_provision_30 = build_provision_30[['building', 'Administrative', 'Munitipal', 'addr:street',
       'addr:housenumber', 'population', 'demand_30', 'demand_left', 'min_dist', 'avg_dist',
       'supplied_demands_within', 'supplied_demands_without', 'provision_value', 'geometry']]

build_provision_30['demand_left'] = build_provision_30['demand_left'].astype('int32')

build_provision_30 = build_provision_30.rename(columns={
    'building': 'Тип здания',
    'Administrative': 'Административный район',
    'Munitipal': 'Муниципальное образование',
    'addr:street': 'Улица',
    'addr:housenumber': 'Номер дома',
    'population': 'Количество жителей',
    'demand_30': 'Спрос в границах временной доступности 30 минут',
    'demand_left': 'Спрос, оставшийся неудовлетворенным',
    'min_dist': 'Наименьшее расстояние до скейт-парка',
    'avg_dist': 'Среднее расстояние до скейт-парка',
    'supplied_demands_within': 'Спрос, удовлетворенный в границах временной доступности 30 минут',
    'supplied_demands_without': 'Спрос, удовлетворенный вне границ временной доступности',
    'provision_value': 'Показатель обеспеченности жилого дома скейт-парками',
    }, inplace=False)

build_provision_30.to_file(os.path.join(data_path,"build_provision_30.geojson"))
build_provision_30.to_file(os.path.join(data_path,"Показатель обеспеченности жилых домов скейт-парками в границах 30 минут.geojson"))
```

```python
# Перемещения между жилыми домами и скейт-парками

distance_between_30 = distance_between_30[['demand', 'distance', 'geometry']]

distance_between_30 = distance_between_30.rename(columns={
    'demand': 'Количество пользователей',
    'distance': 'Время в пути',
    }, inplace=False)

distance_between_30.to_file(os.path.join(data_path,"distance_between_30.geojson"))
distance_between_30.to_file(os.path.join(data_path,"Перемещения между домами и скейт-парками в границах 30 минут.geojson"))
```

```python
# Показатель загруженности скейт-парков

del skateparks_load_30['capacity']
del skateparks_load_30['capacity_left']

skateparks_load_30 = skateparks_load_30.rename(columns={
    'TableID': 'Номер в таблице',
    'Name': 'Название',
    'Rating': 'Рейтинг',
    'Administrative': 'Административный район',
    'Munitipal': 'Муниципальное образование',
    'Address': 'Адрес',
    'Operator': 'Обслуживающая организация',
    'Year': 'Год постройки',
    'Area': 'Площадь постройки',
    'Dimensional_type': 'Размерный тип',
    'Construction_type': 'Тип конструкции',
    'Base_material': 'Материал основания',
    'Terms_of_use': 'Правила пользования',
    'Dimensional_solution': 'Объемно-пространственное решение',
    'Disciplines': 'Подходящие дисциплины',
    'Riding_Styles': 'Подходящие стили катания',
    'Styles_count': 'Количество подходящих стилей',
    'Required_riding_level': 'Требуемый уровень катания',
    'fixed_capacity': 'Вместимость',
    'carried_capacity_within': 'Вместимость, используемая в границах временной доступности 30 минут',
    'carried_capacity_without': 'Вместимость, используемая вне границ временной доступности',
    'service_load': 'Количество посетителей',
    'load_level': 'Показатель уровня загруженности'
    }, inplace=False)

skateparks_load_30.to_file(os.path.join(data_path,"skateparks_load_30.geojson"))
skateparks_load_30.to_file(os.path.join(data_path,"Показатель загруженности скейт-парков в границах 30 минут.geojson"))
```

```python
# Аггрегация данных по муниципальным образованиям

def min_max_normalization(series):
    return (series - series.min()) / (series.max() - series.min())

final_assessment_30['Нормализованный показатель загруженности скейт-парков'] = min_max_normalization(final_assessment_30['Показатель общего уровня загруженности скейт-парков']).fillna(0).round(3)

final_assessment_30 = final_assessment_30.rename(columns={
    'Вместимость скейт-парков, используемая вне границ временной доступности 30 минут': 'Вместимость скейт-парков, используемая вне границ временной доступности',
    'Общая вместимость скейт-парков': 'Суммарная единовременная пропускная способность',
    }, inplace=False)

final_assessment_30 = final_assessment_30[['Административный район',
                                           'Муниципальное образование',
                                           'Население',
                                           'Спрос в границах временной доступности 30 минут',
                                           'Спрос, удовлетворенный в границах временной доступности 30 минут',
                                           'Спрос, удовлетворенный вне границ временной доступности',
                                           'Спрос, оставшийся неудовлетворенным',
                                           'Среднее расстояние до скейт-парка',
                                           'Нормализованный показатель транспортной доступности скейт-парков',
                                           'Показатель обеспеченности населения в районе скейт-парками',
                                           'Количество скейт-парков',
                                           'Суммарная единовременная пропускная способность',
                                           'Вместимость скейт-парков, используемая в границах временной доступности 30 минут',
                                           'Вместимость скейт-парков, используемая вне границ временной доступности',
                                           'Вместимость скейт-парков, оставшаяся неиспользуемой',
                                           'Суммарная нагрузка на скейт-парк',
                                           'Показатель общего уровня загруженности скейт-парков',
                                           'Нормализованный показатель загруженности скейт-парков',
                                           'Показатель уровня загруженности скейт-парков с учетом их количества',
                                           'Итоговая оценка обеспеченности района с учетом количества скейт-парков',
                                           'Итоговая оценка обеспеченности района без учета количества скейт-парков',
                                           'Итоговая оценка обеспеченности района с учетом нормализованного показателя загруженности',
                                           'geometry']]
```

```python
# Показатель обеспеченности населения скейт-парками в муниципальном образовании

provision_30 = final_assessment_30[['Административный район', 'Муниципальное образование',
                                    'Население', 'Количество скейт-парков',
                                    'Спрос в границах временной доступности 30 минут',
                                    'Спрос, удовлетворенный в границах временной доступности 30 минут',
                                    'Спрос, удовлетворенный вне границ временной доступности',
                                    'Спрос, оставшийся неудовлетворенным',
                                    'Показатель обеспеченности населения в районе скейт-парками',
                                    'geometry']]
```

```python
# Показатель транспортной доступности скейт-парков в муниципальном образовании

distance_30 = final_assessment_30[['Административный район',
                                   'Муниципальное образование',
                                   'Население',
                                   'Количество скейт-парков',
                                   'Среднее расстояние до скейт-парка',
                                   'Нормализованный показатель транспортной доступности скейт-парков',
                                   'geometry']]
```

```python
# Показатель загруженности скейт-парков в муниципальном образовании

load_30 = final_assessment_30[['Административный район',
                               'Муниципальное образование',
                               'Население',
                               'Количество скейт-парков',
                               'Суммарная единовременная пропускная способность',
                               'Вместимость скейт-парков, используемая в границах временной доступности 30 минут',
                               'Вместимость скейт-парков, используемая вне границ временной доступности',
                               'Суммарная нагрузка на скейт-парк',
                               'Показатель общего уровня загруженности скейт-парков',
                               'Нормализованный показатель загруженности скейт-парков',
                               'Показатель уровня загруженности скейт-парков с учетом их количества',
                               'geometry']]
```

```python
# Итоговая оценка обеспеченности для муниципальных образований

final_30 = final_assessment_30[['Административный район',
                                'Муниципальное образование',
                                'Население',
                                'Спрос в границах временной доступности 30 минут',
                                'Спрос, удовлетворенный в границах временной доступности 30 минут',
                                'Спрос, удовлетворенный вне границ временной доступности',
                                'Спрос, оставшийся неудовлетворенным',
                                'Показатель обеспеченности населения в районе скейт-парками',
                                'Количество скейт-парков',
                                'Среднее расстояние до скейт-парка',
                                'Нормализованный показатель транспортной доступности скейт-парков',
                                'Суммарная единовременная пропускная способность',
                                'Суммарная нагрузка на скейт-парк',
                                'Показатель общего уровня загруженности скейт-парков',
                                'Нормализованный показатель загруженности скейт-парков',
                                'Показатель уровня загруженности скейт-парков с учетом их количества',
                                'Итоговая оценка обеспеченности района с учетом количества скейт-парков',
                                'Итоговая оценка обеспеченности района без учета количества скейт-парков',
                                'Итоговая оценка обеспеченности района с учетом нормализованного показателя загруженности',
                                'geometry']]
```

```python
# Сохранение файлов с английским названием

provision_30.to_file(os.path.join(data_path,"provision_30.geojson"))
distance_30.to_file(os.path.join(data_path,"distance_30.geojson"))
load_30.to_file(os.path.join(data_path,"load_30.geojson"))
final_30.to_file(os.path.join(data_path,"final_30.geojson"))
```

```python
# Сохранение файлов на русском языке

provision_30.to_file(os.path.join(data_path,"Показатель обеспеченности населения скейт-парками в границах 30 минут.geojson"))
distance_30.to_file(os.path.join(data_path,"Показатель транспортной доступности скейт-парков в границах 30 минут.geojson"))
load_30.to_file(os.path.join(data_path,"Показатель загруженности скейт-парков в границах 30 минут.geojson"))
final_30.to_file(os.path.join(data_path,"Итоговая оценка обеспеченности в границах 30 минут.geojson"))
```

```python
# Сохранение запасного файла со всеми показателями на русском языке

final_assessment_30.to_file(os.path.join(data_path,"Все показатели в границах 30 минут.geojson"))
```
